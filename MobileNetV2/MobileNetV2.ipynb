```python
import numpy as np
import random
import tensorflow as tf

# Reprodutibilidade
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Configurações
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
NUM_CLASSES = 5

# Caminhos
TRAIN_DIR = DATASET / "Treinamento"
VAL_DIR = DATASET / "Validação"
TEST_DIR = DATASET / "Teste"

print("Treinamento:", TRAIN_DIR.exists())
print("Validação:", VAL_DIR.exists())
print("Teste:", TEST_DIR.exists())

print("\nTamanho das imagens:", IMG_SIZE)
print("Batch size:", BATCH_SIZE)
print("Número de classes:", NUM_CLASSES)
print("Seed:", SEED)
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

print("\nClasses:")
print(train_ds.class_names)
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

print("\nClasses reconhecidas:")
print(train_ds.class_names)
```

```python
print("Treino:    ", train_ds.class_names)
print("Validação: ", val_ds.class_names)
print("Teste:     ", test_ds.class_names)

print("\nMesma ordem treino/validação:",
      train_ds.class_names == val_ds.class_names)

print("Mesma ordem treino/teste:",
      train_ds.class_names == test_ds.class_names)
```

```python
CLASS_NAMES = train_ds.class_names

preprocess_input = tf.keras.applications.mobilenet_v2.preprocess_input

train_ds = train_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.prefetch(tf.data.AUTOTUNE)

print("Pré-processamento da MobileNetV2 aplicado com sucesso.")
print("Classes:", CLASS_NAMES)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3))

x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)

outputs = tf.keras.layers.Dense(
    NUM_CLASSES,
    activation="softmax",
    kernel_regularizer=tf.keras.regularizers.l2(1e-4)
)(x)

model = tf.keras.Model(inputs, outputs)

print("MobileNetV2 criada com sucesso.")
print("Base congelada:", not base_model.trainable)
print("Número de classes:", NUM_CLASSES)
```

```python
model.summary()

trainable_params = np.sum(
    [np.prod(v.shape) for v in model.trainable_weights]
)

non_trainable_params = np.sum(
    [np.prod(v.shape) for v in model.non_trainable_weights]
)

print("\nParâmetros treináveis:", trainable_params)
print("Parâmetros não treináveis:", non_trainable_params)
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Modelo compilado com sucesso.")
```

```python
for i, layer in enumerate(base_model.layers[-40:]):
    indice_real = len(base_model.layers) - 40 + i
    print(indice_real, layer.name, type(layer).__name__)
```

```python
base_model.trainable = True

for layer in base_model.layers:
    layer.trainable = False

for layer in base_model.layers:
    if (
        layer.name.startswith("block_14")
        or layer.name.startswith("block_15")
        or layer.name.startswith("block_16")
        or layer.name == "Conv_1"
    ):
        if not isinstance(layer, tf.keras.layers.BatchNormalization):
            layer.trainable = True

print("Fine-tuning configurado.")

print("\nCamadas treináveis da base:")
for layer in base_model.layers:
    if layer.trainable:
        print(layer.name, type(layer).__name__)
```

```python
trainable_params_ft = sum(
    tf.keras.backend.count_params(p)
    for p in model.trainable_weights
)

non_trainable_params_ft = sum(
    tf.keras.backend.count_params(p)
    for p in model.non_trainable_weights
)

print("Parâmetros treináveis no fine-tuning:", trainable_params_ft)
print("Parâmetros não treináveis:", non_trainable_params_ft)
print("Parâmetros totais:", trainable_params_ft + non_trainable_params_ft)
```

```python
LR_FINE_TUNING = 5e-6

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LR_FINE_TUNING
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Modelo recompilado para o fine-tuning.")
print("Learning rate:", LR_FINE_TUNING)
```

```python
CHECKPOINT_FT = RESULTADOS / "melhor_modelo_finetuning.keras"

callbacks_fase2 = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-7,
        verbose=1
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(CHECKPOINT_FT),
        monitor="val_loss",
        mode="min",
        save_best_only=True,
        verbose=1
    )
]

print("Callbacks do fine-tuning configurados.")
print("Checkpoint:", CHECKPOINT_FT)
```

```python
EPOCHS_FASE2 = 16

history_fase2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_FASE2,
    callbacks=callbacks_fase2,
    verbose=1
)
```

```python
melhor_modelo = tf.keras.models.load_model(
    RESULTADOS / "melhor_modelo_finetuning.keras"
)

print("Melhor modelo de fine-tuning carregado com sucesso.")
```

```python
test_loss, test_accuracy = melhor_modelo.evaluate(
    test_ds,
    verbose=1
)

print(f"\nLoss no teste: {test_loss:.4f}")
print(f"Acurácia no teste: {test_accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np

y_true = []
y_pred = []

for imagens, rotulos in test_ds:
    predicoes = melhor_modelo.predict(imagens, verbose=0)

    y_true.extend(np.argmax(rotulos.numpy(), axis=1))
    y_pred.extend(np.argmax(predicoes, axis=1))

y_true = np.array(y_true)
y_pred = np.array(y_pred)

print("Previsões concluídas.")
print("Total de imagens:", len(y_true))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
import unicodedata

# Apenas normaliza os nomes para exibição correta dos acentos
CLASS_NAMES_DISPLAY = [
    unicodedata.normalize("NFC", nome)
    for nome in CLASS_NAMES
]

accuracy = accuracy_score(y_true, y_pred)

precision_macro = precision_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

recall_macro = recall_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1_macro = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

print("Métricas gerais:")
print(f"Acurácia:      {accuracy * 100:.2f}%")
print(f"Precisão macro:{precision_macro * 100:.2f}%")
print(f"Recall macro:  {recall_macro * 100:.2f}%")
print(f"F1-score macro:{f1_macro * 100:.2f}%")

print("\nRelatório por classe:\n")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=CLASS_NAMES_DISPLAY,
        digits=4,
        zero_division=0
    )
)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_true, y_pred)

fig, ax = plt.subplots(figsize=(10, 8))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=CLASS_NAMES_DISPLAY
)

disp.plot(
    ax=ax,
    values_format="d",
    xticks_rotation=45
)

plt.title("Matriz de Confusão - MobileNetV2")
plt.tight_layout()

CAMINHO_MATRIZ = RESULTADOS / "matriz_confusao_mobilenetv2.png"

plt.savefig(
    CAMINHO_MATRIZ,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nMatriz numérica:")
print(cm)

print("\nImagem salva em:")
print(CAMINHO_MATRIZ)
```
