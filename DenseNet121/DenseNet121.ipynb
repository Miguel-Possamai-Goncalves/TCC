```python
total_geral = 0

for divisao in sorted(DATASET_LOCAL.iterdir()):
    if divisao.is_dir():
        print(f"\n{divisao.name}:")

        total_divisao = 0

        for classe in sorted(divisao.iterdir()):
            if classe.is_dir():
                imagens = [
                    arq for arq in classe.iterdir()
                    if arq.suffix.lower() in {".jpg", ".jpeg", ".png"}
                ]

                quantidade = len(imagens)
                total_divisao += quantidade
                total_geral += quantidade

                print(f"  {classe.name}: {quantidade}")

        print(f"  Total da divisão: {total_divisao}")

print(f"\nTotal geral: {total_geral}")
```

```python
import random
import numpy as np
import tensorflow as tf

SEED = 42
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
NUM_CLASSES = 5

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

pastas_local = {
    item.name: item
    for item in DATASET_LOCAL.iterdir()
    if item.is_dir()
}

for nome, caminho in pastas_local.items():
    if nome == "Treino":
        TRAIN_DIR_LOCAL = caminho
    elif nome == "Teste":
        TEST_DIR_LOCAL = caminho
    elif nome.startswith("Valid"):
        VAL_DIR_LOCAL = caminho

print("Treino:", TRAIN_DIR_LOCAL)
print("Validação:", VAL_DIR_LOCAL)
print("Teste:", TEST_DIR_LOCAL)

print("\nTamanho das imagens:", IMG_SIZE)
print("Batch size:", BATCH_SIZE)
print("Número de classes:", NUM_CLASSES)
print("Seed:", SEED)
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR_LOCAL,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR_LOCAL,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR_LOCAL,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

CLASS_NAMES = train_ds.class_names

print("\nClasses reconhecidas:")
print(CLASS_NAMES)
```

```python
preprocess_input = tf.keras.applications.densenet.preprocess_input

train_ds = train_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.prefetch(tf.data.AUTOTUNE)

print("Pré-processamento da DenseNet121 aplicado com sucesso.")
print("Classes:", CLASS_NAMES)
```

```python
base_model = tf.keras.applications.DenseNet121(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3))

x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)

outputs = tf.keras.layers.Dense(
    NUM_CLASSES,
    activation="softmax",
    kernel_regularizer=tf.keras.regularizers.l2(1e-4)
)(x)

model = tf.keras.Model(inputs, outputs)

print("DenseNet121 criada com sucesso.")
print("Base congelada:", not base_model.trainable)
print("Número de classes:", NUM_CLASSES)
```

```python
model.summary()

trainable_params = np.sum(
    [np.prod(v.shape) for v in model.trainable_weights]
)

non_trainable_params = np.sum(
    [np.prod(v.shape) for v in model.non_trainable_weights]
)

print("\nParâmetros treináveis:", trainable_params)
print("Parâmetros não treináveis:", non_trainable_params)
```

```python
print("Base DenseNet121 treinável:", base_model.trainable)

print("\nPesos treináveis do modelo:")
for peso in model.trainable_weights:
    print(peso.name, peso.shape)

print("\nTotal de parâmetros treináveis:")
print(
    sum(
        tf.keras.backend.count_params(p)
        for p in model.trainable_weights
    )
)
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("DenseNet121 compilada com sucesso.")
```

```python
from pathlib import Path

RESULTADOS = TCC / "RESULTADOS" / "DenseNet121"
RESULTADOS.mkdir(parents=True, exist_ok=True)

CHECKPOINT = RESULTADOS / "melhor_modelo_fase1.keras"

callbacks_fase1 = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-7,
        verbose=1
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(CHECKPOINT),
        monitor="val_loss",
        mode="min",
        save_best_only=True,
        verbose=1
    )
]

print("Callbacks configurados com sucesso.")
print("Resultados serão salvos em:", RESULTADOS)
```

```python
EPOCHS_FASE1 = 16

history_fase1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_FASE1,
    callbacks=callbacks_fase1,
    verbose=1
)
```

```python
indices_conv5 = [
    i for i, layer in enumerate(base_model.layers)
    if layer.name.startswith("conv5_")
]

print("Total de camadas da DenseNet121:", len(base_model.layers))
print("Camadas pertencentes ao bloco conv5:", len(indices_conv5))

print("\nPrimeira camada do conv5:")
i = indices_conv5[0]
print(i, base_model.layers[i].name, type(base_model.layers[i]).__name__)

print("\nÚltima camada do conv5:")
i = indices_conv5[-1]
print(i, base_model.layers[i].name, type(base_model.layers[i]).__name__)

print("\nÚltimas 10 camadas da DenseNet121:")
for i, layer in list(enumerate(base_model.layers))[-10:]:
    print(i, layer.name, type(layer).__name__)
```

```python
base_model.trainable = True

# Primeiro congela novamente todas as camadas
for layer in base_model.layers:
    layer.trainable = False

# Libera apenas o último dense block: conv5
for layer in base_model.layers:
    if layer.name.startswith("conv5_"):
        if not isinstance(layer, tf.keras.layers.BatchNormalization):
            layer.trainable = True

print("Fine-tuning da DenseNet121 configurado.")

print("\nCamadas treináveis da base:")
for layer in base_model.layers:
    if layer.trainable:
        print(layer.name, type(layer).__name__)
```

```python
trainable_params_ft = sum(
    tf.keras.backend.count_params(p)
    for p in model.trainable_weights
)

non_trainable_params_ft = sum(
    tf.keras.backend.count_params(p)
    for p in model.non_trainable_weights
)

print("Parâmetros treináveis no fine-tuning:", trainable_params_ft)
print("Parâmetros não treináveis:", non_trainable_params_ft)
print("Parâmetros totais:", trainable_params_ft + non_trainable_params_ft)
```

```python
LR_FINE_TUNING = 5e-6

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LR_FINE_TUNING
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("DenseNet121 recompilada para o fine-tuning.")
print("Learning rate:", LR_FINE_TUNING)
```

```python
CHECKPOINT_FT = RESULTADOS / "melhor_modelo_finetuning.keras"

callbacks_fase2 = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-7,
        verbose=1
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(CHECKPOINT_FT),
        monitor="val_loss",
        mode="min",
        save_best_only=True,
        verbose=1
    )
]

print("Callbacks do fine-tuning configurados.")
print("Checkpoint:", CHECKPOINT_FT)
```

```python
EPOCHS_FASE2 = 16

history_fase2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_FASE2,
    callbacks=callbacks_fase2,
    verbose=1
)
```

```python
best_model = tf.keras.models.load_model(
    RESULTADOS / "melhor_modelo_finetuning.keras"
)

test_loss, test_accuracy = best_model.evaluate(
    test_ds,
    verbose=1
)

print(f"\nTest loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy:.4%}")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_true = []
y_pred = []

for imagens, rotulos in test_ds:
    predicoes = best_model.predict(imagens, verbose=0)

    y_true.extend(np.argmax(rotulos.numpy(), axis=1))
    y_pred.extend(np.argmax(predicoes, axis=1))

y_true = np.array(y_true)
y_pred = np.array(y_pred)

accuracy = accuracy_score(y_true, y_pred)
precision_macro = precision_score(y_true, y_pred, average="macro")
recall_macro = recall_score(y_true, y_pred, average="macro")
f1_macro = f1_score(y_true, y_pred, average="macro")

print(f"Acurácia:        {accuracy:.4%}")
print(f"Precisão macro:  {precision_macro:.4%}")
print(f"Recall macro:    {recall_macro:.4%}")
print(f"F1-score macro:  {f1_macro:.4%}")

print(f"\nAcertos: {np.sum(y_true == y_pred)}")
print(f"Erros:   {np.sum(y_true != y_pred)}")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

class_names = [
    "Ferrugem-asiática",
    "Mancha-alvo",
    "Mancha-olho-de-rã",
    "Mancha-púrpura",
    "Saudável"
]

relatorio = classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    digits=4
)

matriz = confusion_matrix(y_true, y_pred)

print("RELATÓRIO DE CLASSIFICAÇÃO\n")
print(relatorio)

print("MATRIZ DE CONFUSÃO\n")
print(matriz)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(8, 7))

disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=class_names
)

disp.plot(
    ax=ax,
    cmap="Blues",
    values_format="d",
    colorbar=False
)

plt.title("Matriz de Confusão - DenseNet121")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

caminho_matriz = RESULTADOS / "matriz_confusao_densenet121.jpg"

plt.savefig(
    caminho_matriz,
    format="jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Matriz salva em:")
print(caminho_matriz)
```
